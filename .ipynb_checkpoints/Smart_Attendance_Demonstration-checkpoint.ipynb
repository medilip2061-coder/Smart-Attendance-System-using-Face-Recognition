```python

# Read attendance records from the database
conn = sqlite3.connect(database_path)

attendance_df = pd.read_sql_query("""
    SELECT
        a.date,
        a.time,
        s.student_id,
        s.name,
        a.status
    FROM attendance AS a
    LEFT JOIN students AS s
        ON a.student_id = s.student_id
    ORDER BY a.date DESC, a.time DESC
""", conn)

conn.close()

print("ATTENDANCE RECORDS")

if attendance_df.empty:
    print("No attendance records found yet.")
else:
    display(attendance_df)
    print("Total attendance records:", len(attendance_df))
```

## 5. Attendance Analysis and Visualization

This section analyzes attendance records and visualizes
student attendance information using Python and Matplotlib.

```python
# Attendance statistics

import matplotlib.pyplot as plt
from IPython.display import display

print("ATTENDANCE ANALYSIS")

if attendance_df.empty:
    print("No attendance data available yet.")
else:
    total_records = len(attendance_df)
    present_count = (
        attendance_df["status"].str.lower() == "present"
    ).sum()

    absent_count = (
        attendance_df["status"].str.lower() == "absent"
    ).sum()

    attendance_rate = (
        present_count / total_records * 100
    )

    print("Total attendance records:", total_records)
    print("Present records:", present_count)
    print("Absent records:", absent_count)
    print(f"Attendance rate: {attendance_rate:.2f}%")
```

```python
# Attendance visualization

if not attendance_df.empty:
    status_counts = (
        attendance_df["status"]
        .str.title()
        .value_counts()
    )

    status_counts.plot(
        kind="bar",
        figsize=(7, 4),
        title="Student Attendance Overview",
        xlabel="Attendance Status",
        ylabel="Number of Records"
    )

    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()
```

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

# 1. Point to your captured faces dataset folder (e.g., 'dataset' or 'training_images')
dataset_path = r"C:\Users\1mdil\Desktop\Smart_Attendance_System Face Recognisation\dataset"

student_counts = {}
sample_images = []
sample_labels = []

# Scan folders/images in your dataset
for root, dirs, files in os.walk(dataset_path):
    img_files = [f for f in files if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
    if img_files:
        # Use subfolder name or filename prefix as student label
        label = os.path.basename(root) if root != dataset_path else "Registered_Students"
        student_counts[label] = len(img_files)
        
        # Grab 1 sample face per student for visual grid
        img_path = os.path.join(root, img_files[0])
        img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
        if img is not None:
            sample_images.append(img)
            sample_labels.append(label)

# 2. Plot 1: Bar Chart of Training Images per Student
plt.figure(figsize=(8, 4))
plt.bar(student_counts.keys(), student_counts.values(), color='#2b5b84')
plt.title("LBPH Face Recognition - Training Samples per Student", fontsize=12, fontweight='bold')
plt.xlabel("Student ID / Folder")
plt.ylabel("Number of Face Images")
plt.xticks(rotation=30)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# 3. Plot 2: Visual Grid of Preprocessed Grayscale Training Faces
if sample_images:
    cols = min(len(sample_images), 5)
    fig, axes = plt.subplots(1, cols, figsize=(12, 3))
    if cols == 1:
        axes = [axes]
    for ax, img, lbl in zip(axes, sample_images[:cols], sample_labels[:cols]):
        ax.imshow(img, cmap='gray')
        ax.set_title(lbl, fontsize=9)
        ax.axis('off')
    plt.suptitle("Sample Grayscale Faces Used for LBPH Model Training", fontweight='bold')
    plt.tight_layout()
    plt.show()
```
